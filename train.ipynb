```python
import pandas as pd
import numpy as np
```

```python
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
data = pd.read_csv('data/final_data.csv')
X = data.drop(columns=['Placement'])
y = data['Placement']
```

```python
# Z-Score

upper_limit = X['IQ'].mean() + 3*X['IQ'].std()
lower_limit = X['IQ'].mean() - 3*X['IQ'].std()
```

```python
# Handling Outliers (Capping)

X['IQ'] = np.where(
    X['IQ'] > upper_limit,
    upper_limit,
    
    np.where(
        X['IQ'] < lower_limit,
        lower_limit,
        X['IQ']
    )
)
```

```python
ct1 = ColumnTransformer([
    ('trf1', PolynomialFeatures(degree=2), X.columns), 
    ('trf2', StandardScaler(), X.columns)
], remainder='passthrough')
```

```python
lr = LogisticRegression()
```

```python
pipe = Pipeline([
    ('step-1', ct1), 
    ('step-2', lr)
])
```

```python
pipe.fit(X, y)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score

accuracy_score(y, pipe.predict(X))
np.mean(cross_val_score(pipe, X, y, scoring='accuracy'))
```

```python
import joblib

joblib.dump(pipe, 'placement_predictor.pkl')
```
